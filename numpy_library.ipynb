creating 1d array

```
a=np.array([1,2,3,4,5])
print(a)
```

```
#2d array
b=np.array([(1,2,3,4),(5,6,7,8)])
print(b)
```

initial placeholders in numpy arrays

```
#create a numpy array of zeroes
x=np.zeros((4,5))
print(x)
```

```
#create a numpy array of 1s
y=np.ones((3,3))
print(y)
```

```
#array of a particular value
p=np.full((5,5),5)
print(p)
```

```
#create an identity matrix
a=np.eye(4)
print(a)
```

```
#create a numpy array of random values btw 0 and 1
b=np.random.random((3,4))
print(b)
```

```
#random integer value of the array within a specific range
c= np.random.randint(10,100,(3,5))   # range=(10,100), dimension =3,5
print(c)
```

```
#array of evenly spaced values , specifying the no of values
d= np.linspace(10,30,5) # no of values=5
print(d)
```

```
 #array of evenly spaced values , mentioning the step size
 e= np.arange(10,30,5)   #step size =5
 print(e)
```

```
#convert a list to an array
lst2= [1,2,3,4,5]

np_array=np.asarray(lst2)
print(np_array)
type(np_array)
```

```
#array dimension
print(c.shape)
```

```
# no of dimensions
print(c.ndim)
```

mathematical operation on np array


```
lst1=[1,2,3,4,5]
lst2=[5,6,7,8,9]
print(lst1+lst2)
#concatenation on adding
```

```
a=np.random.randint(0,10,(3,3))
b=np.random.randint(10,20,(3,3))
print(a)
print(b)
```

```
print(a+b)
print(a-b)
print(a*b)
print(a/b)
```

```
print(np.add(a,b))
print(np.subtract(a,b))
```

array manipulation

```
#transpose
trans=np.transpose(a)
print(trans)
print(a.shape)
```

```
b=np.random.randint(10,20,(3,2))
print(b)
trans=np.transpose(b)
print(trans)
print(b.shape)
print(trans.shape)
```

```
#reshaping an array
a=np.random.randint(0,10,(2,3))
print(a)
print(a.shape)
```

```
b=a.reshape(3,2)
print(b)
```
